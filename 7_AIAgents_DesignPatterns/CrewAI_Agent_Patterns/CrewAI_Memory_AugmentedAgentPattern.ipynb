# Define the Memory-Aware Agent

```python
# Define the Memory-Aware Agent
chat_agent = Agent(
    role="Conversational Memory Agent",
    goal="Engage in conversation while remembering prior context",
    backstory="You are a helpful assistant who maintains memory throughout a conversation.",
    tools=[],
    verbose=True,
    allow_delegation=False,
    llm=ChatOpenAI(temperature=0)
)
```

# Task Template (simulates memory turn-by-turn)

```python
# ---
# Task Template (simulates memory turn-by-turn)
def create_task(input_text: str, memory: str):
    return Task(
        description=f"Prior memory: {memory}\nUser message: {input_text}\nRespond conversationally using memory if helpful.",
        expected_output="Conversational response that references memory if applicable.",
        agent=chat_agent
    )
```

# Simulate Memory Across Turns

```python
# ---
# Simulate Memory Across Turns
conversation = []
user_turns = [
    "Hi, who won the World Cup in 2018?",
    "Remind me what we just talked about.",
    "Now give me a fact about the winning country."
]

for user_input in user_turns:
    memory_text = "\n".join([f"User: {u}\nAgent: {a}" for u, a in zip(conversation[::2], conversation[1::2])])
    task = create_task(user_input, memory_text)
    crew = Crew(tasks=[task], agents=[chat_agent], verbose=True)
    result = crew.kickoff()
    conversation.append(user_input)
    conversation.append(result)
    print(f"\n🧠 User: {user_input}\n🤖 Agent: {result}")
```
