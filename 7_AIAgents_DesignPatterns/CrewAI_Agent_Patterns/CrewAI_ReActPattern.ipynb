```python
os.environ["OPENAI_API_KEY"] = "sk---"
```

# Define ReAct Agent

```python
llm = ChatOpenAI(temperature=0)

react_agent = Agent(
    role="ReAct Agent",
    goal="Answer questions by reasoning and using tools",
    backstory="You use intermediate reasoning steps and decide when to apply tools such as search or calculator.",
    tools=[],
    verbose=True,
    llm=llm
)
```

# Define ReAct Task

```python
react_task = Task(
    description="You are asked: 'What is the square root of 144 plus the capital of France?' Reason through this step by step, and decide if you need to use a calculator or search tool.",
    expected_output="Provide a final answer using reasoning and tool use, e.g., calculator or search as needed.",
    agent=react_agent
)
```

# Run the ReAct Crew

```python

print("\n🔁 Running CrewAI ReAct Pattern")
Crew(agents=[react_agent], tasks=[react_task]).kickoff()
```
