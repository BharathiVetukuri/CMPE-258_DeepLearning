# Define Agents

```python
llm = ChatOpenAI(temperature=0)

react_agent = Agent(
    role="ReAct Agent",
    goal="Use reasoning and tool use to answer questions",
    backstory="You select and apply tools using intermediate reasoning.",
    tools=[],
    verbose=True,
    llm=llm
)

tool_selector_agent = Agent(
    role="Tool Selector",
    goal="Select the best tool for a given question",
    backstory="You analyze a user's query and route to the right tool.",
    tools=[],
    verbose=True,
    llm=llm
)

planner = Agent(
    role="Planner",
    goal="Decompose tasks into steps",
    backstory="You are an expert planner who breaks down goals clearly.",
    tools=[],
    verbose=True,
    llm=llm
)

executor = Agent(
    role="Executor",
    goal="Execute steps and solve the task",
    backstory="You complete tasks methodically using given plans.",
    tools=[],
    verbose=True,
    llm=llm
)

summarizer = Agent(
    role="Summarizer",
    goal="Summarize technical content for end users",
    backstory="You make results easier to understand and communicate.",
    tools=[],
    verbose=True,
    llm=llm
)

researcher = Agent(
    role="Researcher",
    goal="Collect information and findings",
    backstory="You dig deep into facts and gather relevant data.",
    tools=[],
    verbose=True,
    llm=llm
)

engineer = Agent(
    role="Engineer",
    goal="Design a technical solution",
    backstory="You translate research into technical implementation.",
    tools=[],
    verbose=True,
    llm=llm
)

reviewer = Agent(
    role="Reviewer",
    goal="Review and critique solutions",
    backstory="You ensure proposed solutions are robust and well thought-out.",
    tools=[],
    verbose=True,
    llm=llm
)
```

# Multi-Agent Collaboration Tasks

```python
# Multi-Agent Collaboration Tasks
research_task = Task(
    description="Research: 'How can AI help with food security?'",
    expected_output="List of relevant findings",
    agent=researcher
)

engineering_task = Task(
    description="Propose an AI-based solution for food security based on research",
    expected_output="Detailed technical concept",
    agent=engineer
)

review_task = Task(
    description="Critically review the proposed AI solution for food security",
    expected_output="Balanced critique",
    agent=reviewer
)
```

```python
!pip install -q crewai openai --upgrade
!pip install -q langchain_community --upgrade
```

# Run Crew

```python
print("\n🔁 Running Multi-Agent Collaboration")
Crew(agents=[researcher, engineer, reviewer], tasks=[research_task, engineering_task, review_task]).kickoff()
```
