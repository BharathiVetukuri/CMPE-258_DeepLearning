# Define Workflow Agents

```python
llm = ChatOpenAI(temperature=0)

planner = Agent(
    role="Planner",
    goal="Decompose tasks into steps",
    backstory="You are an expert planner who breaks down goals clearly.",
    tools=[],
    verbose=True,
    llm=llm
)

executor = Agent(
    role="Executor",
    goal="Execute steps and solve the task",
    backstory="You complete tasks methodically using given plans.",
    tools=[],
    verbose=True,
    llm=llm
)

summarizer = Agent(
    role="Summarizer",
    goal="Summarize technical content for end users",
    backstory="You make results easier to understand and communicate.",
    tools=[],
    verbose=True,
    llm=llm
)
```

# Define Workflow Tasks

```python
# ---
# Define Workflow Tasks
plan_task = Task(
    description="Break down the task: 'Explain the impact of AI on healthcare'",
    expected_output="Clear plan with 2-3 steps",
    agent=planner
)

execute_task = Task(
    description="Carry out the plan to explain the impact of AI on healthcare",
    expected_output="Detailed response per plan",
    agent=executor
)

summary_task = Task(
    description="Summarize the execution output for a non-technical reader",
    expected_output="Short summary",
    agent=summarizer
)
```

# Run the Workflow Crew

```python

print("\n🔁 Running CrewAI Multi-Step Workflow")
Crew(agents=[planner, executor, summarizer], tasks=[plan_task, execute_task, summary_task]).kickoff()
```
