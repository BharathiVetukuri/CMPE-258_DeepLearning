```python
os.environ["OPENAI_API_KEY"] = "sk---"
```

# Define Tool Selector Agent

```python
llm = ChatOpenAI(temperature=0)

tool_selector_agent = Agent(
    role="Tool Selector",
    goal="Choose the correct tool based on the user's query",
    backstory="You intelligently choose between a calculator and a search tool depending on the task.",
    tools=[],
    verbose=True,
    llm=llm
)
```

# Define Tool Selector Task

```python
tool_selection_task = Task(
    description="The user asked: 'What's the capital of Germany plus 2 + 2?'. Decide whether this should go to a search tool or calculator.",
    expected_output="Specify the tool: either 'search_web' or 'calculator' and give the appropriate answer.",
    agent=tool_selector_agent
)
```

# Run the Tool Selector Crew

```python
# ---
# Run the Tool Selector Crew
print("\n🔁 Running CrewAI Tool Selector Pattern")
Crew(agents=[tool_selector_agent], tasks=[tool_selection_task]).kickoff()
```
